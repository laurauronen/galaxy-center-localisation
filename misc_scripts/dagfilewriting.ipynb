```python
import numpy as np 
import os
```

```python
start_index = 0
total = 100
jobs = 10
allbkg = 10000

for i in range(0,int(allbkg/(jobs*total))):
    title = f'small_bkg_dag_{i}.dag'
    with open(title, 'w') as f:
        for j in range(0,jobs):
            f.write(f'JOB job_{j:03d} /home/lauraelina.uronen/GClocalization/background_submit_small.sub\n')
            f.write(f'VARS job_{j:03d} start="{start_index}" cpus="20" total="{total}" job="{j:03d}" outdir="/home/lauraelina.uronen/GClocalization/small_system/batch_{i}/" log_tag="/home/lauraelina.uronen/GClocalization/small_system/logs_{i}/job_{j:03d}" \n')
            f.write('\n')
            start_index +=  total
    f.close()
    dir = f'/home/lauraelina.uronen/GClocalization/small_system/logs_{i}/'
    outdir = f'/home/lauraelina.uronen/GClocalization/small_system/batch_{i}/'
    if not os.path.exists(outdir):
        os.makedirs(outdir)
    if not os.path.exists(dir):
        os.makedirs(dir)
```
