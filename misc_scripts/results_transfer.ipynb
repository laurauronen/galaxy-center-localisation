```python
import pickle as pkl
import json 
import os
import numpy as np

indir = '/home/lauraelina.uronen/GClocalization/average_system_rrel/'
outdir = '/home/lauraelina.uronen/GClocalization/galaxy-center-localisation/results_final/rrel/'
```

```python
real_event_path = indir + 'Outdir_lens_reconstruction/true_position.json'

with open(real_event_path) as f:
    real_event = json.load(f)

y_rel_90 = real_event['90th_percentile']
percentile90 = []
n = 0 

for j in range(0,10):
    for i in range(0,100):
        title = indir + f'batch_{j}/writefile_{i}.json'
        if os.path.exists(title):
            data = json.load(open(title))
            for key in data.keys():
                percentile90.append(data[key])
                n += 1

percentile90 = np.array(percentile90).flatten()
prob = np.sum(percentile90 < y_rel_90) / len(percentile90)

output = {'true_90th_percentile': y_rel_90,
          'all_90th_percentiles': percentile90.tolist(),
          'probability': prob}

with open(outdir + 'results.json', 'w') as f:
    json.dump(output, f)
```
